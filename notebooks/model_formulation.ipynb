### Basics

- agents $\mathcal{N}=\{1,\cdots,N\}$ agent $i$
- tasks $\mathcal{T}=\{1,\cdots,T\}$ task $j$

- energy budget of agent $i$ is $B_i$, $\forall i \in \mathcal{N}$
- energy requirement of task $j$ is $R_j$, $\forall j \in \mathcal{T}$

- an assignment is $\mathcal{X} = (\mathcal{X}_{ij})_{i \in \mathcal{N}, j \in \mathcal{T}} \in \mathbb{N}^{N \times T}$

### Assumptions

- Assumption 0 (Connectivity): All members within a team are connected to each other

- Assumption 1 (Feasibility): The total energy budget of all agents is no less than the total energy requirement of all tasks：$\sum_{i \in \mathcal{N}}B_i \ge \sum_{j \in \mathcal{T}}R_j$ 

- Assumption 2 (Minimum energy requirement): $ \min_j R_j > 1 $

- Assumption seems not needed - (Full budget utilisation): Each agent is assumed to allocate its entire available budget, leading to the following constraint $\sum_{j \in \mathcal{T}}\mathcal{X}_{ij} = B_i, \forall i \in \mathcal{N}$



### Definitions

- Definition 1 (Surplus tasks $ \mathcal{T}_s $): Surplus tasks are tasks whose assigned energy exceeds its requirement: $\mathcal{T}_s = \{ j \in \mathcal{T} | \sum_{i \in \mathcal{T}}\mathcal{X}_{ij} > R_j \} $

- Definition 2 (Reassignable agents): An agent is reassignable w.r.t. task $j$ if $ \exists j \in T_i = \{ j \in \mathcal{T} | \mathcal{X}_{ij} > 0 \} $, and $\sum_{i^{\prime} \in \mathcal{N}  / i } \mathcal{X}_{i^{\prime} j} \ge R_j$.

### Hypergraph representation

- an assignment can be mapped to a weighted hypergraph $\mathcal{H} = (\mathcal{V},\mathcal{E},\omega,\theta)$, where $\omega: \mathcal{E} \to \mathbb{N}$ is a hyperedge weight function and $\theta: \mathcal{V} \times \mathcal{E} \to \mathbb{N}$ is a incidence weight function
- agents mapped to a set of nodes $\mathcal{V}$
- tasks mapped to a set of hyperedges $\mathcal{E}$
- energy requirement of task $j$ mapped to the weight $ \omega(e_j) $ of hyperedge $e_j$
- energy that agent $i$ spends in task $j$ mapped to the weight of incidence pair $\theta(v_i,e_j)$
- a solution mapped to a weighted incidence matrix

### Problem Formulation 1

Objective: **persistence**: the ability of the initial assignment to maintain task feasibility ($ \pi(\mathcal{H},p) $)  and connectivity ($ \eta(\mathcal{H},p) $) immediately after perturbations

- disruption type: random incidence-level (i, j)
- resilience measurement: the number of feasible tasks (functional persistence)

$\text{Minimize}\quad \pi(\mathcal{H},p)=\mathbb{E}\left[\sum_j \mathbb{1}_{\{\sum_i \mathcal{X}'_{ij}<R_j\}}\right]$

 $\text{subject to}$

$\sum_j \mathcal{X}_{ij}\leq B_i,\quad \forall i\in N$,

$\sum_i \mathcal{X}_{ij}\geq R_j,\quad \forall j\in T$,

$\mathcal{X}'_{ij}=\mathcal{X}_{ij}(1-\mathcal{Z}_{ij}),\quad \forall i\in N,\ j\in T$,

$\mathcal{X}_{ij}\in\mathbb{N},\quad \mathcal{Z}_{ij}\overset{\text{i.i.d.}}{\sim}\operatorname{Bernoulli}(p),\quad \forall i\in N,\ j\in T$.

### Hypergraph discovery problem w.r.t. Problem Formulation 1

- maximizing the sum of generalized degrees

$\max\ \pi(\mathcal{H})$

$
\pi(\mathcal{H}) = \sum_{v_i} \sum_{e_j} \mathbb{1}_{ \{ \theta(v_i,e_j) > 0 \} }
$

$
\sum_{e_j}\theta(v_i,e_j)\le \beta(v_i),\ \forall v_i\in \mathcal{V},
$

$
\sum_{v_i}\theta(v_i,e_j)\ge w(e_j),\ \forall e_j\in \mathcal{E},
$

### Problem Formulation 2

Objective: **recoverability**: the ease with which disrupted tasks can be restored by reallocating agents with redundant resources

$ \delta(\mathcal{H},p) $: patching costs

- disruption type: random incidence-level (i, j)
- resilience measurement: worst-case recovery burden

$\text{Minimize}\quad 
F_2(\mathcal{X})
=
\mathbb{E}\left[
\max_{\mathcal{Y}\in\mathcal{F}(\mathcal{Z})}
\sum_i\sum_j
\left|
\mathcal{Y}_{ij}-\mathcal{X}_{ij}
\right|
\right]$

$\text{subject to}$

$\sum_j \mathcal{X}_{ij}\leq B_i,\quad \forall i\in\mathcal{N}$,

$\sum_i \mathcal{X}_{ij}\geq R_j,\quad \forall j\in\mathcal{T}$,

$\mathcal{F}(\mathcal{Z})
=
\left\{
\mathcal{Y}\in \mathbb{N}^{N \times T} \ \middle|\ 
\sum_i \mathcal{Y}_{ij}(1-\mathcal{Z}_{ij})\geq R_j,\ \forall j\in\mathcal{T};
\quad
\sum_j \mathcal{Y}_{ij}\leq B_i,\ \forall i\in\mathcal{N}
\right\}$,

$\mathcal{X}_{ij}\in\mathbb{N},\quad
\mathcal{Y}_{ij}\in\mathbb{N},\quad
\mathcal{Z}_{ij}\overset{\text{i.i.d.}}{\sim}\operatorname{Bernoulli}(p),
\quad \forall i\in\mathcal{N},\ j\in\mathcal{T}$.

### Hypergraph discovery problem w.r.t. Problem Formulation 2

- minimizing the length of walks between disrupted tasks and reallocating agents

$\text{Minimize}\quad F_2(\mathcal{X})=\mathbb{E}\left[ \omega^{*}  \right]$

$\text{subject to}$

$\sum_j \mathcal{X}_{ij}\leq B_i,\quad \forall i\in \mathcal{N},$

$\sum_i \mathcal{X}_{ij}\geq R_j,\quad \forall j\in \mathcal{T}$,

$\omega^{*} = \underset{\omega_{ik} \in \Omega_{ik}}{\max} \left| \omega_{ik} \right|, \forall i \in \mathcal{N_n},\; \forall k \in \mathcal{N_p},$,

$\mathcal{N_n}=\left\{i\in \mathcal{N}\mid \sum_j \mathcal{X}_{ij}\mathcal{Z}_{ij}>0\right\}$,

$\mathcal{T_p}=\left\{j\in \mathcal{T}\mid \sum_i \mathcal{X}_{ij}\mathcal{Z}_{ij}>0\right\}$,

$\mathcal{N_p}=\left\{i\in \mathcal{N}\mid \mathcal{X}_{ij}>0,\ j\in \mathcal{T_p}\right\}$,

$\mathcal{X}_{ij}\in\mathbb{N},\quad
\mathcal{Z}_{ij}\overset{\text{i.i.d.}}{\sim}\operatorname{Bernoulli}(p),
\quad \forall i\in \mathcal{N},\ j\in \mathcal{T}$.

### Relaxation of the resilient team formation problem

- constrained simulated annealing $ \mathcal{H}^{\text{SA}} $, 50% greedy repair + 50% random perturbation

$\text{Minimize}\ f(\mathcal{X}) = \sum_{i} \left( \sum_{j} X_{ij} - B \right)^{+} + \sum_{j} \left(R - \sum_{i} X_{ij} \right)^{+} $

$\text{subject to}$

$ \frac{\sum_i \sum_j X_{ij}}{|\mathcal{N}|} \leq B, $

$ \frac{\sum_j \sum_i X_{ij}}{|\mathcal{T}|} \geq R, $

$ X_{ij}\in\{0,1\}, i \in \mathcal{N},\ j \in \mathcal{T}$

- $ \mu_2 $-guided constrained simulated annealing $ \mathcal{H}^{\text{SA}}_{\mu_2} $, 50% greedy repair + 50% random perturbation

$\text{Minimize}\ f(\mathcal{X}) = \sum_{i} \left( \sum_{j} X_{ij} - B \right)^{+} + \sum_{j} \left(R - \sum_{i} X_{ij} \right)^{+} - \mu_2(\mathcal{X}) $

$\text{subject to}$

$ \frac{\sum_i \sum_j X_{ij}}{|\mathcal{N}|} \leq B, $

$ \frac{\sum_j \sum_i X_{ij}}{|\mathcal{T}|} \geq R, $

$ X_{ij}\in\{0,1\}, i \in \mathcal{N},\ j \in \mathcal{T}$
